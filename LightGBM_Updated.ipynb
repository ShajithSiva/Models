```python
import pandas as pd
import numpy as np

from lightgbm import LGBMRegressor

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error


from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Load Dataset
df = pd.read_excel(
    "Expense_Dataset_With_NextMonthExpense.xlsx"
)

# Features & Target
X = df.drop(
    "NextMonthExpense",
    axis=1
)

y = df["NextMonthExpense"]

# Categorical Columns
categorical_features = [
    "User_ID",
    "Spending_Profile"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ],
    remainder="passthrough"
)

# LightGBM Model
model = LGBMRegressor(
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    random_state=42
)

# Pipeline
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", model)
])

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train
pipeline.fit(
    X_train,
    y_train
)

# Predict
predictions = pipeline.predict(
    X_test
)

# Metrics
mae = mean_absolute_error(
    y_test,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predictions
    )
)

r2 = r2_score(
    y_test,
    predictions
)

mape = np.mean(
    np.abs(
        (y_test - predictions)
        / y_test
    )
) * 100

mse = mean_squared_error(y_test, predictions)



print("\nModel Performance")
print("-" * 40)
print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")
print(f"MAPE : {mape:.2f}%")
print(f"MSE : {mse:.2f}")
```
